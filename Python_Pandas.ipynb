#### 1. Index-Based vs. Label-Based Selection (`loc` vs. `iloc`)
*   **`loc` (Label-based):** Selects data using row and column text labels.
*   **`iloc` (Index-based):** Selects data using zero-based integer index positions.

| Operation | `loc[]` (By Label) | `iloc[]` (By Integer Index) |
| :--- | :--- | :--- |
| **Single Row** | `df.loc['r2']` | `df.iloc[1]` |
| **Single Column** | `df.loc[:, 'c2']` | `df.iloc[:, 1]` |
| **Multiple Rows** | `df.loc[['r2', 'r3']]` | `df.iloc[[1, 2]]` |
| **Multiple Columns** | `df.loc[:, ['c2', 'c3']]` | `df.iloc[:, [1, 2]]` |
| **Row Range** | `df.loc['r1':'r4']` *(inclusive)* | `df.iloc[0:4]` *(exclusive stop)* |
| **Column Range** | `df.loc[:, 'c1':'c4']` *(inclusive)* | `df.iloc[:, 0:4]` *(exclusive stop)* |
| **Alternate Rows** | `df.loc['r1':'r5':2]` | `df.iloc[0:5:2]` |
| **Alternate Columns** | `df.loc[:, 'c1':'c5':2]` | `df.iloc[:, 0:5:2]` |
| **Conditional Filter** | `df.loc[df['Fee'] >= 1000]` | `df.iloc[list(df['Fee'] >= 1000)]` |
| **Lambda Function** | `df.loc[lambda x:x[3]]` | `df.iloc[lambda x:x[3]]` |

#### 2.1 GroupBy Process: The Split-Apply-Combine Strategy
The `groupby()` function categorizes data based on custom keys using Hadley Wickham's **Split-Apply-Combine** workflow:

1.  **Split:** Breaks the DataFrame into smaller subsets based on unique key values.
2.  **Apply:** Applies an aggregation function to each independent group.
3.  **Combine:** Combines the individual aggregated results back into a single structured output array.
```
Input DataFrame     Split by Key      Apply Function
┌─────┬──────┐      ┌───┬───┐         ┌───┬───┐
│ Key │ Data │      │ A │ 1 │ ───┐    │ A │ 5 │      Combine Output
├─────┼──────┤      │ A │ 4 │    │    └───┴───┘      ┌───┬──────┐
│  A  │  1   │      └───┴───┘    │                   │ A │  5   │
│  B  │  2   │      ┌───┬───┐    │    ┌───┬───┐      ├───┼──────┤
│  C  │  3   │ ───> │ B │ 2 │ ───┼──> │ B │ 7 │ ───> │ B │  7   │
│  A  │  4   │      │ B │ 5 │    │    └───┴───┘      ├───┼──────┤
│  B  │  5   │      └───┴───┘    │                   │ C │  9   │

                    └───┴───┘

#### 2.2 Pivot Table

A multi-dimensional aggregation and reshaping tool that reorganizes flat tabular data into a two-dimensional grid. It maps specified keys to the row index (y-axis) and column headers (x-axis), calculating summary statistics across their intersections.
```

┌───┬──────┬─────┬─────┬─────┐     ┌──────┬───────┬───────┐

#### 2.3 Cross-Tabulation

A specialized statistical matrix (contingency table) designed to compute frequency distributions and relative percentages across categorical or discrete variables. It analyzes how two or more categorical factors interact.

* **Frequency Table:** Shows the distribution of values in a dataset by counting the number of occurrences (or relative percentages) of unique categorical values within a dataset.
```
Input Lists                        pd.crosstab(index=List_1, columns=List_2, values=List_3, aggfun=sum)
┌────────┬─────────┬────────┐      ┌────────┬────────┬─────────┐
│ List_1 │ List_2  │ List_3 │      │ List_2 │ Inside │ Outside │
├────────┼─────────┤────────┤      ├────────┼────────┼─────────┤
│ Bob    │ Inside  │   10   │      │ List_1 │        │         │
│ Bob    │ Outside │   15   │      ├────────┼────────┼─────────┤
│ Sally  │ Outside │   20   │ ───> │ Bob    │   40   │   15    │
│ Bob    │ Inside  │   30   │      ├────────┼────────┼─────────┤
└────────┴─────────┴────────┘      │ Sally  │    0   │   20    │
                                   └────────┴────────┴─────────┘
```

#### 3. Handling Missing Data (`None` vs. `NaN`)
Pandas represents missing entries using two values:
*   `None`: Python's built-in singleton object.
*   `NaN` (Not a Number): IEEE floating-point representation.
Pandas treats `None` and `NaN` as interchangeable null values.

```
               ┌─ Checking Nulls ─────┬─ isnull() (True if missing)
               │                      └─ notnull() (True if present)
               │
               ├─ Imputing / Filling ─┬─ fillna(), ffill(), bfill()
Missing Data ──┤                      ├─ replace()
  Handling     │                      └─ interpolate()
               │
               └─ Dropping Values ────── dropna()
```
```
                ┌─────────────────────────────────┐
                │ Methods to Handle Missing Value │
                └────────────────┬────────────────┘
       ┌─────────────────────────┴────────────────────┐
┌──────┴─────┐                                  ┌─────┴────┐
│ Imputation │                                  │ Deletion │
└───┬────────┘                                  └─────┬────┘
    ├─ Mean ─── Numerical data, normal distribution   └── Use when: < 5% missing data
    ├─ Median ─ Numerical data, skewed distribution   & no meaningful pattern
    └─ Mode ─── Categorical data (e.g., "Weather")    (MCAR - Missing Completely At Random)
```

##### Categorical Data & Feature Encoding
Machine learning models require numerical input, necessitating categorical encoding.

*   **Nominal Encoding:** Discrete categories with no inherent ordering (e.g., states, colors).
*   **Ordinal Encoding:** Discrete categories with an explicit rank or order. Assigns an integer (0, 1, 2...) to each unique category.

##### Encoding Techniques
*   **One-Hot Encoding:** Maps nominal values to binary vectors (0 or 1).
    *   *Dummy Variable Trap:* To prevent perfect multicollinearity, one column must be dropped, keeping (N-1) dummy columns.
    *   *Curse of Dimensionality:* Applying One-Hot Encoding to features with high cardinality inflate dataset dimensionality.
*   **Label / Map Encoding:** Maps ordered categories directly to incremental integers reflecting their rank.

````python
import pandas as pd
z = pd.DataFrame({'Size':['Small', 'Medium', 'Large']})
size_map = {'Small': 0, 'Medium': 1, 'Large': 2}
z['Size'].map(size_map)
````

````python
z = pd.DataFrame({'Color':['Red', 'Blue', 'Green']})
pd.get_dummies(z, columns=['Color'])
````

##### Statistical Functions
*   `pct_change()`: Calculates percentage change from a previous element to the current element.

DataFrame.rolling(window, min_periods=None, center=False, win_type=None, on=None, axis=0, closed=right, step=None)

* window: If offset string (e.g., '2D', '5h', '10s'): Time-based window based on timestamps (requires DatetimeIndex or on parameter).
* min_periods: Minimum observations required in the accumulating window before producing a non-NaN output.
* win_type: Weight distribution across window points (requires SciPy).
* on: Name of a column containing datetimes to use for time-based rolling instead of using the DataFrame index.
* closed: • 'right' (T-N, T] • 'left' [T-N, T) • 'both' [T-N, T] • 'neither' (T-N, T) $(for time-based window)$
* step: Evaluates the window every $N$-th step instead of every single row.

````python
df4 = pd.DataFrame({'Date': pd.date_range(start='2026-01-01', periods=5, freq='D'), 'Sales':[100, 120, 110, 150, 200]}).set_index('Date')
df4['Rolling_3D_Mean'] = df4['Sales'].rolling(window=3).mean()
df4['Rolling_3D_Mean_Center'] = df4['Sales'].rolling(window=3, center=True).mean()
df4['Rolling_3D_Min1'] = df4['Sales'].rolling(window=3, min_periods=1).mean() # (to prevents leading NaNs)
print(df4)
````

````python
df5 = pd.DataFrame({"X": 1}, index=[pd.Timestamp("20130101 09:00:01"), pd.Timestamp("20130101 09:00:02"), pd.Timestamp("20130101 09:00:03")])
df5["right"] = df5['X'].rolling("2s", closed="right").sum()
df5["both"] = df5.rolling("2s", closed="both").X.sum() # Other Syntax
df5["left"] = df5.rolling("2s", closed="left").X.sum()
df5["neither"] = df5.rolling("2s", closed="neither").X.sum()
print(df5)
````

DataFrame.rolling().apply(func, raw=False, engine='python', engine_kwargs=None, args=(), kwargs={})

* func: A custom Python function that accepts an array/Series representing the current window and returns a single numerical scalar value.
* raw: • True: Passes a ndarray • False: Passes a pandas Series
* engine: 'python' or 'numba'
> engine_kwargs, args, kwargs ✅

````python
import numpy as np
def custom_range(arr):
    return np.max(arr) - np.min(arr)

df4['Rolling_Range'] = df4['Sales'].rolling(window=3).apply(custom_range, raw=True)
df4[['Sales', 'Rolling_Range']]
````

````python
def weighted_mean(arr, w1, w2, w3):
    return (arr[0]*w1 + arr[1]*w2 + arr[2]*w3) / (w1 + w2 + w3)

df4['Weighted_Rolling'] = df4['Sales'].rolling(3).apply(weighted_mean, raw=True, args=(1, 2, 3))  # Pass extra weights via args=(w1, w2, w3)
df4[['Sales', 'Weighted_Rolling']]
````

DataFrame.expanding(min_periods=1, axis=0, method='single')

````python
df4['Cumulative_Sum'] = df4['Sales'].expanding(min_periods=2).sum()
df4[['Sales', 'Cumulative_Sum']]
````

DataFrame.ewm(com=None, span=None, halflife=None, alpha=None, min_periods=0, adjust=True, ignore_na=False)

* com: Center of mass ($\alpha = \frac{1}{1 + \text{com}}$).
* span: $N$-period moving average decay ($\alpha = \frac{2}{N + 1}$). Most commonly used.
* halflife: Time for weight to decay by half.
* alpha: Direct smoothing factor ($0 < \alpha \le 1$).
> adjust, ignore_na ✅

````python
df4['EMA_3'] = df4['Sales'].ewm(span=3).mean()
df4[['Sales', 'EMA_3']]
````

##### **Row Numbers:** Row numbering assigns a sequential index (0, 1, 2...) to rows.

##### Quantile Binning

`pd.qcut()` discretizes a continuous numeric variable into **equal-sized frequency buckets** (quantiles), ensuring each bucket contains approximately the same number of samples.

pd.qcut(x, q, labels=None, retbins=False, precision=3, duplicates='raise')

* q: Number of quantiles (e.g., 4) or array of quantiles [0, 0.5, 0.9, 1.0].
* duplicates: `'raise'` or `'drop'`. If bin edges overlap due to duplicate values, `'drop'` merges identical bins.
> retbins, precision ✅

##### Lag & Lead Operations
It allow you to compare a row's current value with past or future rows without changing the DataFrame order.

* Lag (Looking Back): Shifts data downward, bringing previous values into the current row.
* Lead (Looking Ahead): Shifts data upward, bringing future values into the current row.

Series.shift(periods=1, freq=None, axis=0, fill_value=np.nan)

* freq: Shifts the Datetime Index instead of shifting row values.
* fill_value: Value used to replace boundary NaN.

When working with grouped data (groupby), you often need to pick specific positional records per group (e.g., a customer's first purchase).

DataFrame.groupby('group_column').nth(n, dropna=None)
* nth(0) $\rightarrow$ First value in group.
* nth(-1) $\rightarrow$ Last value in group.

* groupby().first(): Returns the first non-null value for each column within each group.
* groupby().last(): Returns the last non-null value for each column within each group.

````python
df4['Lag_1'] = df4['Score'].shift(-1)
df4['Lead_1'] = df4['Score'].shift(1)
df4[['Score', 'Lag_1', 'Lead_1']]
````

````python
df4['nth_first'] = df4.groupby('Department')['Score'].agg('nth', 0) # Should be `transform`
df4['nth_last'] = df4.groupby('Department')['Score'].agg('nth', -1) # Should be `transform`
df4['non_null_first'] = df4.groupby('Department')['Score'].transform('first')
df4['non_null_last'] = df4.groupby('Department')['Score'].transform('last')
df4[['Department', 'Score', 'nth_first', 'nth_last', 'non_null_first', 'non_null_last']]
````
